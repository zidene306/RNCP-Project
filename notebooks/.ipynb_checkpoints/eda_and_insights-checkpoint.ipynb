```python
#-----------------------------------------------------------------------------
# Do regions with larger populations tend to have different 5G performance?
#-----------------------------------------------------------------------------

#-----------------------------------------------------------------------------
# 1. JOINING Population data with geo data: pop_geo_merge_df(population_df, geo_df)
#-----------------------------------------------------------------------------
#open Geographic file

geo_df_file_path = os.path.join(clean_folder, "insee_geo.csv")
geo_df = pd.read_csv(geo_df_file_path, sep=",", dtype={"insee_com": "str"}, encoding="latin1")

#open population file
population_df_file_path = os.path.join(clean_folder, "insee_com_population.csv")
population_df = pd.read_csv(population_df_file_path, sep=",", dtype={"insee_com": "str"}, encoding="latin1")

pop_geo_clean_df = fn.pop_geo_merge_df(population_df, geo_df)
denormalized_qos_file_name = "denormalized_qos.csv"

denormalized_qos_df = pd.read_csv(os.path.join(clean_folder, "denormalized_qos.csv"), sep=",", dtype={"insee_com": "str", "insee_dep": "str",}, encoding="latin1")
denormalized_qos_df = denormalized_qos_df.drop(columns="reg_name")
qos_normalized_df = denormalized_qos_df.merge(pop_geo_clean_df[["insee_com", "population", "reg_name"]], how="left", on="insee_com")
qos_normalized_df.columns
```

```python
#-----------------------------------------------------------------------------
# 2. Study the downloadspeed vs infra, population, radio conditions
#-----------------------------------------------------------------------------
df_perf_population = qos_normalized_df.groupby(["protocol_code", "reg_name", "operator_name"]).agg(
    average_download_speed=("bitrate_dl", "mean"),
    average_upload_speed=("bitrate_ul", "mean"),
    total_phys_sites=("tot_phys_sites", "sum"),
    total_pure_5g_sites=("tot_5g_per_op_com", "sum"),
    average_rsrp=("rsrp", "mean"),
    average_rsrq=("rsrq", "mean"),   
    population=("population", "sum")).reset_index()

df_perf_population.head()
df_pop_down_speed = df_perf_population[df_perf_population["protocol_code"] == "DOWNLOAD"]
df_pop_up_speed = df_perf_population[df_perf_population["protocol_code"] == "UPLOAD"]

#SAve files
df_pop_up_speed.to_csv(os.path.join(clean_folder, "up_speed_region_radio_cond.csv"), index=False, encoding="utf-8")
df_pop_down_speed.to_csv(os.path.join(clean_folder, "down_speed_region_radio_cond.csv"), index=False, encoding="utf-8")
    
df_pop_up_speed.head()
```

```python
#-----------------------------------------------------------------------------
# 3. Hypothesis testing
#-----------------------------------------------------------------------------

# 3.1 Do operators have different 5G download speeds?
# H₀: the average 5G download speed is the same for all operators.
# H ₁: at least one operator differs.

from scipy.stats import f_oneway
from scipy.stats import spearmanr

orange_cond = df_pop_down_speed[df_pop_down_speed["operator_name"] == "Orange"]
orange = orange_cond[orange_cond["protocol_code"] == "DOWNLOAD"]["average_download_speed"]

sfr_cond = df_pop_down_speed[df_pop_down_speed["operator_name"] == "SFR"]
sfr = sfr_cond[sfr_cond["protocol_code"] == "DOWNLOAD"]["average_download_speed"]

bytel_cond = df_pop_down_speed[df_pop_down_speed["operator_name"] == "Bouygues Telecom"]
bytel = bytel_cond[bytel_cond["protocol_code"] == "DOWNLOAD"]["average_download_speed"]

free_cond = df_pop_down_speed[df_pop_down_speed["operator_name"] == "Free Mobile"]
free = free_cond[free_cond["protocol_code"] == "DOWNLOAD"]["average_download_speed"]

F_stat, p_value = f_oneway(
    orange,
    bytel,
    sfr,
    free)

print("F-statistic:", F_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print(f"p_value = {p_value} < 0.05. Reject H0: there is no siginifacant evidence that all the 4 operators have the same average download speed.")
else:
    print(f"p_value {p_vlaue} > 0.05. Cannot reject H0. There is significant evidence that the 4 operators have the same average download speed.")
```

```python
# 3.2 Population vs. 5G download speed: Spearmann correlation
# H₀: There is no significant association between regional population and average 5G download speed.
# H₁: There is a significant association between regional population and average 5G download speed.

df_population = df_pop_down_speed[df_pop_down_speed["protocol_code"] == "DOWNLOAD"][["reg_name", "operator_name", "average_download_speed", "population"]]
df_population

spear_stat, p_value = spearmanr(df_population["population"], 
                                df_population["average_download_speed"]
                               )
print("spear_stat:", spear_stat)
print("p_value:", p_value)

if p_value <0.05:
    print("Reject H0: There is statistically significant evidence that there is a positive association between population and average 5G download speed.")
else:
    print("Fail to reject H0: There is statistically significant evidence that there is a positive association between population and average 5G download speed.")
        
```

```python
# 3.3 Population vs. 5G download speed: Spearmann correlation operator by operator
# H₀: There is no significant association between RSRP and average 5G download speed on operator level
# H₁: There is a significant association between RSRP and average 5G download speed of  operator level
operator_list = df_pop_down_speed["operator_name"].unique()
for operator in operator_list:
    print(f"\n========={operator}===========")
    stat, p_value = spearmanr(df_pop_down_speed[df_pop_down_speed["operator_name"] == operator]["population"],
        df_pop_down_speed[df_pop_down_speed["operator_name"] == operator]["average_download_speed"])
    if p_value < 0.05:
        print(f"stat, p_value_tuple=({stat}, {p_value}) for {operator} and p_value < 0.05: Reject H0. There is statistically significant evidence that population and average down speed are associated.")
    else:
        print(f"stat, p_value_tuple=({stat}, {p_value}) for {operator}and p_value > 0.05: fail to reject H0. There is statistically significant evidence that population and average down speed are not associated.")
    
```

```python
# 3.4 RSRP vs. 5G download speed: Spearmann correlation
# H₀: There is no significant association between RSRP and average 5G download speed.
# H₁: There is a significant association between RSRP and average 5G download speed.

df_rsrp = df_pop_down_speed[
    ["reg_name", "operator_name", "average_rsrp", "average_download_speed"]
].dropna()

from scipy.stats import spearmanr

rho, p_value = spearmanr(
    df_rsrp["average_rsrp"],
    df_rsrp["average_download_speed"]
)

print(f"Spearman rho: {rho:.3f}")
print(f"p-value: {p_value:.6f}")
```

```python
for operator, group in df_rsrp.groupby("operator_name"):
    rho, p_value = spearmanr(
        group["average_rsrp"],
        group["average_download_speed"]
    )

    print(f"{operator}:")
    print(f"  Spearman rho = {rho:.3f}")
    print(f"  p-value = {p_value:.6f}")
```

```python
# 3.2 infrastructure vs. 5G download speed: Spearmann correlation
# H₀: There is no significant association between regional nbr of sites and average 5G download speed.
# H₁: There is a significant association between regional nbr of sites and average 5G download speed.

df_population = df_pop_down_speed[df_pop_down_speed["protocol_code"] == "DOWNLOAD"][["reg_name", "operator_name", "average_download_speed", "total_pure_5g_sites"]]
df_population

spear_stat, p_value = spearmanr(df_population["total_pure_5g_sites"], 
                                df_population["average_download_speed"]
                               )
print("spear_stat:", spear_stat)
print("p_value:", p_value)

if p_value <0.05:
    print("Reject H0: There is statistically significant evidence that there is a positive association between nbr of 5G sites and average 5G download speed.")
else:
    print("Fail to reject H0: There is statistically significant evidence that there is a positive association between nbr of 5G sites and average 5G download speed.")
```

```python
# 3.3 nbr of 5G sites vs. 5G download speed: Spearmann correlation operator by operator
# H₀: There is no significant association between nbr of 5G sites and average 5G download speed on operator level
# H₁: There is a significant association between nbr of 5G sites and average 5G download speed of  operator level
operator_list = df_pop_down_speed["operator_name"].unique()
for operator in operator_list:
    print(f"\n========={operator}===========")
    stat, p_value = spearmanr(df_pop_down_speed[df_pop_down_speed["operator_name"] == operator]["total_pure_5g_sites"],
        df_pop_down_speed[df_pop_down_speed["operator_name"] == operator]["average_download_speed"])
    if p_value < 0.05:
        print(f"stat, p_value_tuple=({stat}, {p_value}) for {operator} and p_value < 0.05: Reject H0. There is statistically significant evidence that nbr of 5G sites and average down speed are associated.")
    else:
        print(f"stat, p_value_tuple=({stat}, {p_value}) for {operator}and p_value > 0.05: fail to reject H0. There is statistically significant evidence that nbr of 5G sites and average down speed are not associated.")
    
```
