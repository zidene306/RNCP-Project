```python
from flask import Flask, jsonify, request
import os
from dotenv import load_dotenv

import mysql.connector

```

```python
#Establish connection to SQL DATABASE

def get_connection():

    load_dotenv()

    user = os.getenv("DB_USER")
    password = os.getenv("DB_PASSWORD")
    host = os.getenv("DB_HOST")
    port = os.getenv("DB_PORT")
    db = os.getenv("DB_NAME")

    return mysql.connector.connect(
        user=user,
        password=password,
        port=port,
        host=host,
        database=db
    )
```

```python
# Enpoint1: "/api/qos" get qos by commune and operator

app = Flask(__name__)

@app.route("/api/qos", methods=["GET"])
def get_com_qos():

    # Request parameters
    operator = request.args.get("operator")
    insee_com = request.args.get("insee_com")

    # Check that both parameters are provided
    if operator and insee_com:

        query = """
            SELECT measure_id,
                   insee_com,
                   operator_name,
                   protocol_code,
                   bitrate_dl,
                   bitrate_ul,
                   rsrp,
                   rsrq
            FROM qos_denormalized
            WHERE operator_name = %s
              AND insee_com = %s
            LIMIT 100
        """

        conn = get_connection()
        cursor = conn.cursor(dictionary=True)
        cursor.execute(query, (operator, insee_com))
    
    elif operator or insee_com:

        return jsonify({
            "error": "Please provide both operator and insee_com"
        }), 400
    
 
    else:

        query = """
            SELECT measure_id,
                   insee_com,
                   operator_name,
                   protocol_code,
                   bitrate_dl,
                   bitrate_ul,
                   rsrp,
                   rsrq
            FROM qos_denormalized
            LIMIT 100
        """
      

        conn = get_connection()
        cursor = conn.cursor(dictionary=True)

        cursor.execute(query)

    data = cursor.fetchall()

    cursor.close()
    conn.close()

    return jsonify(data)


if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```

```python
# Enpoint2: "/api/communal_stats" get communal statistics: 

app = Flask(__name__)
@app.route("/api/communal_stats", methods=["GET"])
def get_com_qos_stats():

    operator_name = request.args.get("operator_name")
    insee_com = request.args.get("insee_com")

    if not operator_name or not insee_com:
        return jsonify({
            "error": "Please provide both operator_name and insee_com"
        }), 400

    conn = get_connection()
    cursor = conn.cursor(dictionary=True)

    query = """
    SELECT
        %s AS operator_name,
        %s AS insee_com,

        (
            SELECT AVG(bitrate_dl)
            FROM qos_denormalized
            WHERE operator_name = %s
              AND insee_com = %s
              AND protocol_code = 'DOWNLOAD'
        ) AS average_download_speed,

        (
            SELECT AVG(bitrate_ul)
            FROM qos_denormalized
            WHERE operator_name = %s
              AND insee_com = %s
              AND protocol_code = 'UPLOAD'
        ) AS average_upload_speed,

        (
            SELECT MAX(tot_5g_per_op_com)
            FROM qos_denormalized
            WHERE operator_name = %s
              AND insee_com = %s
        ) AS number_5g_sites
    """

    cursor.execute(query, (
        operator_name, insee_com,
        operator_name, insee_com,
        operator_name, insee_com,
        operator_name, insee_com
    ))

    data = cursor.fetchall()

    cursor.close()
    conn.close()

    return jsonify(data)


if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```

```python
# Enpoint3: "/api/regional_qos" get regional QoS

app = Flask(__name__)
@app.route("/api/regional_qos", methods=["GET"])
def get_regional_qos_stats():

    operator_name = request.args.get("operator_name")
    reg_name = request.args.get("reg_name")

    if not operator_name or not reg_name:
        return jsonify({
            "error": "Please provide both operator_name and reg_name"
        }), 400

    conn = get_connection()
    cursor = conn.cursor(dictionary=True)

    query = """
        SELECT
            operator_name,
            reg_name,

            AVG(
                CASE
                    WHEN protocol_code = 'DOWNLOAD'
                    THEN bitrate_dl
                END
            ) AS average_download_speed,

            AVG(
                CASE
                    WHEN protocol_code = 'UPLOAD'
                    THEN bitrate_ul
                END
            ) AS average_upload_speed

        FROM qos_denormalized

        WHERE operator_name = %s
          AND reg_name = %s

        GROUP BY operator_name, reg_name
    """

    cursor.execute(query, (operator_name, reg_name))

    data = cursor.fetchall()

    cursor.close()
    conn.close()

    return jsonify(data)

if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```
